### 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы
* Рассмотрим событие $A$ - варианты распределения призов, если по каждой номинации установлены: **различные призы**

$P(A) = \overline{A_n^k} = n^k$ = $10^5$ = 100 000

```python
result = pow(10,5)
print(result)
```

* Рассмотрим событие $B$ – получилось слово если по каждой номинации установлены: а) **одинаковые призы**

$\tilde{C}_{10}^5 = C_{10+5-1}^5 = C_{14}^5 = \frac{14 \cdot 13 \cdot 12 \cdot 11 \cdot 10}{1 \cdot 2 \cdot 3 \cdot 4 \cdot 5} = 2002$

```python
result = (14*13*12*11*10) / (1*2*3*4*5)
print(result)
```

### 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

> Каждое семизначное число отличается от друтого
порядком следования цифр (причем 1 = 3, n2 = 2, п3 = 2, а их сумма равна 7), т.е. является перестановкой с повторениями из 7 элементов.

$P_7(3; 2; 2) = \frac{7!}{3! \, 2! \, 2!} = 210$

```python
import scipy.special as sc
result = (sc.factorial(7)) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
print(result)
```

### 1.10
>Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?
* Рассмотрим событие $A$ - получение слова "ТОР"

$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$

```python
round(1 / (6*5*4),4)
```

* Рассмотрим событие $B$ - получение слова "ТЕОРИЯ"

$P(B) = \frac{m}{n} = \frac{1}{A_6} = \frac{1}{6!} = \frac{1}{720}$

```python
import scipy.special as sc
round( (1 / sc.factorial(6)) , 4)
```

### 1.11

карточках написаны три буквы А, две буквы Н и одна буква С.

* Пусть событие В - получение слова «АНАНАС»

$P(B) = \frac{m}{n} = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \, 2!}{6!} = \frac{1}{60}$

### 1.16
> За круглым столом рассаживаются 5 мужчин и
5 женщин. Найти вероятность того, что: а) никакие два лица одного
пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

* Пусть событие $А$ - никакие два лица одного
пола не сядут рядом.

$P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!} = \frac{2 \cdot 120^2}{3\,628\,800} = 0{,}00794$

```python
round((2*pow(120,2)) / 3628800 , 4)
```

* Пусть событие $В$ - супруги, образующие пять супружеских
пар, сядут рядом.

$P(B) = \frac{m_2}{n} = \frac{5! \cdot 2^5}{10!} = \frac{120 \cdot 32}{3\,628\,800} = 0{,}00106$

```python
round( (120*32) / (3628800) , 4)
```
